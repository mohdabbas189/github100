```python
#ARGS AND KWARGS

## *args->we can pass multiple arguments at a time ->o/p comes as a list
## **kwargs-> we can pass multiple arguments but we need to mention key(variable name )-->o/p comes as dictionary.
```

```python
def f5(*args):
    print(args)
    print(type(args))
f5(10,20,30,40,50,"abbas")    
```

```python
def f6(**kwargs):
    print(kwargs)
    print(type(kwargs))
f6(a=10,b=40,c=30,d=19)    
```
